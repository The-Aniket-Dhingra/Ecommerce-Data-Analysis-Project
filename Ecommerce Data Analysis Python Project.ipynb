### Take the log of quantity

```python
# Taking the log of quantity
data['log_qty'] = np.log(data['quantity'])
data.head()
```

### Create funtion to identify outliers

```python
# Creating the function which helps us to identify the outliers in the data
def remove_qty_outlier(group):
    q1 = group.quantile(0.25)
    q3 = group.quantile(0.75)
    iqr = q3 - q1
    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr

    return (group < lower_limit) | (group > upper_limit)
```

#### Identify outliers by their respective sub category

```python
# Create column is_outliers which is boolean which tells us is the data outlier by per sub category
data['is_outlier'] = data.groupby('sub_category')['log_qty'].transform(remove_qty_outlier)
```

### Take the log of price and Create funtion to identify outliers

```python
# Taking the log of price
data_dropped['log_price'] = np.log(data_dropped['price'])

# Creating the function which helps us to identify outliers in the data
def remove_price_outliers(group):
    q1 = group.quantile(0.25)
    q3 = group.quantile(0.75)
    iqr = q3 - q1
    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr

    return (group < lower_limit) | (group > upper_limit)
```

#### Identify outliers by their respective sub category

```python
# Create column is_price_outlier which is boolean which tells us is the data oultier by per sub category
data_dropped['is_price_outlier'] = data_dropped.groupby('sub_category')['log_price'].transform(remove_price_outliers)
data_dropped.head()
```
